```python
import numpy as np
import pandas as pd
```

# **1. DataFrames Basic Operations**

```python
df1 = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [100, 200, 300, 400, 500]
})
```

# **2. Column Operations**

**Access Column**

```python
df1['A']
```

**Multiple Columns**

```python
df1[['A', 'B']]
```

**Add New Column**

```python
df1['E'] = df1['A'] + df1['B']
df1['E']
```

**Delete Column**

```python
df1.drop('E', axis=1, inplace=True)
print(df1)
```

# **3. Row Operations**

**Access Row (by index)**

```python
df1.loc[0]
```

**Access Multiple Rows**

```python
df1.loc[0:2]
```

**Access by Position**

```python
df1.iloc[0]
```

# **4. Filtering Data**

**Single Condition**

```python
df1[df1['A'] > 2]
```

**Multiple Conditions**

```python
df1[(df1['A'] > 2) & (df1['B'] < 50)]
```

# **5. Sorting**

**Sort by One Column**

```python
df1.sort_values(by='B')
```

**Descending Order**

```python
df1.sort_values(by='B', ascending=False)
```

# **6. Applying Functions**

```python
df1['D'] = df1["B"].apply(lambda x: x**2)
df1['D']
```

**Apply on Multiple Columns**

```python
df1[['A', 'B']].apply(lambda x: x + 1)
```

```python
df1['B'].mean()
```

# **8. GroupBy**

```python
df2 = pd.DataFrame({
    'A': ['X', 'Y', 'X', 'Y', 'X'],
    'B': [10, 20, 30, 40, 50]
})
df2
```

**Group and Sum**

```python
df2.groupby('A')['B'].sum()
```
